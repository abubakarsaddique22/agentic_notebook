# 0️⃣ Pehle Mental Model: Agent asal mein hai kya?

🎯 **Ek line mein:** Agent = **LLM (dimagh) + Tools (haath paon) + Loop (bar bar sochna-karna) + State/Memory (notebook) + Guardrails (hudood)**.

### 🧠 Naye intern wali misaal

Socho aap ki dukaan mein ek naya intern aaya:

- **Sirf LLM call** = intern se ek sawal poocha, us ne zubaani jawab de diya. Na computer khola, na register dekha.
- **Workflow** = aap ne intern ko likh kar diya: *"Pehle order number lo → system mein dekho → customer ko status batao."* Woh har bar yahi karta hai. Raasta **aap** ne tay kiya.
- **Agent** = aap ne intern ko kaha: *"Customer ka masla hal karo."* Ab woh khud sochta hai: pehle order dekhun? stock check karun? manager se poochun? Raasta **intern khud** tay karta hai — magar sirf un cheezon se jo aap ne use di hain (tools), aur un hudood mein jo aap ne lagayi hain (permissions).

### Sab se aham baat
> **Agent ki pehchan yeh hai ke "agla qadam kya hoga" yeh faisla model karta hai, aap ka code nahi.**

Baqi sab — tools, memory, frameworks — isi ek idea ke gird ki cheezen hain.

### Kyun 2026 mein yeh sab se demand wala hunar hai?
- Companies ab "chatbot jo baat kare" nahi, balki "system jo **kaam kar de**" chahti hain — order process kare, ticket banaye, report likhe, code theek kare.
- Iske liye LLM ko bahar ki duniya se jorna (tools, MCP), kaam yaad rakhna (memory), aur ghalti hone par sambhalna (reliability) seekhna parta hai. Yahi Level 5 hai.

# 5.1 What is an Agent? — Agent aur Workflow ka farq

🎯 **Ek line mein:** Workflow mein raasta **code** tay karta hai; agent mein raasta **model** tay karta hai. Har masle ko agent nahi chahiye.

## 🧠 Definition

**AI Agent:** Aisa system jahan LLM ek goal ke liye **khud** tay karta hai ke kaunsa tool chalaye, nateeja dekhe, aur phir agla qadam kya ho — jab tak kaam mukammal na ho jaye ya rukne ki shart poori na ho.

Anthropic ki mashhoor guide *"Building Effective Agents"* yeh farq aise karti hai:
- **Workflows:** LLMs aur tools ko **pehle se likhe code ke raaston** se chalaya jata hai.
- **Agents:** LLM **khud apna process aur tools ka istemal dynamically** direct karta hai.

Dono ko mila kar "**agentic systems**" kehte hain.

## 🔁 Agent Loop

```
         ┌──────────────── Goal / User ki request ────────────────┐
         ▼                                                        │
   ┌───────────┐      ┌──────────────┐      ┌────────────────┐    │
   │  DECIDE   │ ───► │  ACT         │ ───► │  OBSERVE       │    │
   │ (kya karun│      │ (tool chalao)│      │ (nateeja dekho)│    │
   └───────────┘      └──────────────┘      └────────────────┘    │
         ▲                                          │             │
         └──────────── dobara socho (loop) ─────────┘             │
                               │                                  │
                      kaam ho gaya? / limit?                      │
                               ▼                                  │
                      ┌──────────────────┐                        │
                      │ FINISH: jawab do │ ───────────────────────┘
                      └──────────────────┘
```

**Goal → Decide → Act → Observe → Repeat → Finish**

## 🪜 Autonomy ki seerhi: Call → Chain → Workflow → Agent

Interview mein yeh table zubani yaad honi chahiye. Farq sirf ek sawal ka hai: **"agla qadam kaun tay karta hai?"**

| Level | Naam | Agla qadam kaun tay karta hai | Watch store example |
|---|---|---|---|
| 0 | **Single LLM call** | Koi nahi (ek hi qadam) | Customer review ka sentiment batana |
| 1 | **Chain** | Code (fixed tarteeb) | Message ka summary → phir Urdu mein translate |
| 2 | **Router** | LLM sirf *raasta* chunta hai, phir code | Sawal "billing" hai ya "delivery" ya "exchange"? |
| 3 | **Workflow + tools** | Code flow, LLM sirf steps bharta hai | Order # nikalo → DB se status lao → reply likho |
| 4 | **Agent** | **LLM (loop mein)** | "Ghari kharab aayi, size bhi chhota hai, address bhi badalna hai" |
| 5 | **Multi-agent** | Kayi LLMs mil kar | Research system: planner + researchers + writer |

> 💡 Jitna upar jao: **flexibility barhti hai**, magar **cost, latency, unpredictability aur debugging ka dard** bhi barhta hai.

## 🧩 Agentic Patterns (Anthropic ki guide se — interview favourite)

Yeh "workflow" patterns hain jo aksar poore agent se behtar hote hain:

| Pattern | Kya hai | Kab istemal |
|---|---|---|
| **Prompt Chaining** | Kaam ko fixed steps mein toro; har step ka output agle ka input; beech mein checks (gates) | Kaam saaf hisson mein bat sakta ho |
| **Routing** | Input classify karo, phir specialized prompt/model ko bhejo | Alag qism ke sawal alag handling maangte hon |
| **Parallelization** | (a) *Sectioning:* kaam ke hisse ek saath; (b) *Voting:* ek hi kaam kayi bar, phir majority | Raftar chahiye, ya zyada bharosa chahiye |
| **Orchestrator–Workers** | Ek LLM kaam ko dynamically baant'ta hai, workers karte hain, phir woh jorta hai | Subtasks pehle se pata na hon |
| **Evaluator–Optimizer** | Ek LLM banata hai, doosra check kar ke feedback deta hai, loop | Wazeh quality criteria ho (translation, code) |
| **Autonomous Agent** | Poora loop model ke haath | Open-ended, raasta pehle se tay na ho |

## 🧱 Agent ke Components

| Component | Kaam | Intern wali misaal |
|---|---|---|
| **Model** | Sochna aur faisla karna | Intern ka dimagh |
| **Instructions (system prompt)** | Role, hudood, policy, style | Job description |
| **Tools** | Bahar ki duniya mein action / maloomat | Computer, register, phone |
| **State / Memory** | Abhi kaam kahan hai, kya yaad rakhna hai | Intern ki notebook |
| **Control loop** | Decide-act-observe ka chakkar | Kaam karne ka routine |
| **Stopping conditions** | Kab rukna hai | "Kaam ho jaye ya 30 minute guzar jayein to batao" |
| **Guardrails / permissions** | Kya karne ki ijazat nahi | "Refund manager ki approval ke baghair nahi" |

## ⚖️ Faisla: Agent chahiye ya Workflow?

**Agent tab jab (zyada tar "haan"):**
- Steps pehle se pata nahi; har case alag raasta maangta hai
- Kaam open-ended hai (research, debugging, complex support)
- Model ke paas **feedback** hai (tool results, tests) jis se woh apni ghalti pakar sake
- Thori zyada **latency aur cost** qabool hai
- Ghalti ka nuqsan **mehdood** hai, ya ghalti pakarne ka nizaam (approval, sandbox) mojood hai

**Workflow tab jab:**
- Steps fixed aur maloom hain (80% support sawal yahi hote hain)
- Compliance / audit: har bar ek hi raasta chahiye
- High volume, kam latency, kam cost
- Ghalti mehngi hai aur raasta predictable rakhna zaroori

> 🔑 **Golden rule:** *Sab se simple cheez se shuru karo jo kaam kar de.* Pehle single LLM call, phir workflow, aur agent **sirf tab** jab naap kar pata chale ke simple tareeqa fail ho raha hai.

## 🏭 Real-world: Watch store ka WhatsApp bot

| Customer ka sawal | Kya lagayenge | Kyun |
|---|---|---|
| "Delivery charges kitne hain?" | Single call + policy context | Seedha jawab, koi action nahi |
| "Order #1043 kahan hai?" | **Workflow** (extract → lookup → reply) | Raasta hamesha ek jaisa |
| "COD available hai?" | Router → FAQ | Fixed jawab |
| "Ghari ka strap toot gaya, exchange chahiye, aur address bhi badal gaya hai" | **Agent** (limited tools) | Kayi faisle, tarteeb case par munhasir |
| "Refund chahiye" | Agent + **human approval** | Paisa = na palatne wala action |

Yani production mein aksar **hybrid** hota hai: **router** aage, aasan cases workflow ko, mushkil cases agent ko.

## 📉 Compounding Error — interview ka sab se taqatwar point

Agar agent ka har qadam **95%** sahi ho, to:

| Steps | Poora kaam sahi hone ka chance |
|---|---|
| 1 | 95% |
| 5 | ~77% |
| 10 | ~60% |
| 20 | ~36% |

(0.95 ko steps ki power do.) Is liye lambe agents ke saath **har qadam ki reliability, checkpoints, aur verification** itni zaroori hai. Yeh ek line interview mein aap ko "production-minded" dikhati hai.

## 🎤 Interview Q&A — 5.1

**Q1. Agent aur workflow mein kya farq hai?**
> Farq control flow ka hai. Workflow mein agla qadam mera code tay karta hai — raasta fixed aur predictable. Agent mein LLM loop mein khud decide karta hai kaunsa tool chalana hai aur kab rukna hai. Workflow sasta, tez aur reliable; agent flexible magar mehnga aur kam predictable. Main default workflow se shuru karta hoon aur agent tab lagata hoon jab raasta pehle se tay na ho sake.

**Q2. Aap agent kab NAHI banayenge?**
> Jab steps maloom hon, volume zyada ho, latency kam chahiye ho, ya compliance ke liye har bar ek hi raasta chahiye. Misal: order status check — yeh 3 fixed steps hain, agent lagana faltu cost aur risk hai.

**Q3. Kya ChatGPT ek agent hai?** *(tricky)*
> Sirf chat ki shakal mein nahi — woh ek LLM call hai. Lekin jab woh tools (search, code execution, browsing) loop mein khud chunta aur chalata hai, tab woh agentic behavior hai. Agent hona model ki khoobi nahi, **system design** ki khoobi hai.

**Q4. Agent ke components batayein.**
> Model, instructions, tools, state/memory, control loop, stopping conditions, aur guardrails/permissions. (Har ek ki ek line mein wazahat do.)

**Q5. Production agents ka sab se bara challenge kya hai?**
> Reliability. Errors compound hoti hain — 95% per-step accuracy bhi 10 steps mein ~60% end-to-end success deti hai. Is liye steps kam rakhna, har tool ko mazboot banana, verification aur human checkpoints lagana, aur trajectory-level evals karna zaroori hai.

**Q6. "Agentic" aur "agent" mein farq?**
> "Agentic" ek spectrum hai — jitna zyada control model ke paas, utna zyada agentic. Router thora agentic hai, autonomous loop poora agent.

## ⚠️ Aam galtiyan
- Har cheez ko agent bana dena kyunke "trend" hai
- Demo mein 5 sawal chala kar production-ready samajh lena
- Stopping conditions aur budgets ke baghair agent chalana
- Agent ko zaroorat se zyada tools aur permissions de dena

## ✅ Yaad rakhne wali baat
> **"Agla qadam kaun tay karta hai?"** — isi sawal se call, chain, workflow aur agent alag hote hain. Simple se shuru karo; complexity naap kar barhao.

# 5.2 ReAct, Planning & Reflection — Agent sochta aur karta kaise hai

🎯 **Ek line mein:** ReAct = **socho → karo → dekho → phir socho**; Planning = pehle naqsha banao; Reflection = apne kaam ko jaanch kar sudharo.

## 🧠 ReAct (Reason + Act)

2022 ke paper (*Yao et al.*) se aaya. Idea simple hai: model sirf sochta nahi (Chain-of-Thought) aur sirf karta nahi (Act-only) — dono **bari bari** karta hai, aur har action ke **nateeje (observation)** ko dekh kar agla faisla karta hai.

### Ek ReAct trace (text mein)

```
User: Order #1043 kahan hai aur kab tak aayega?

Thought 1: Mujhe order ka status chahiye. get_order tool istemal karta hoon.
Action 1:  get_order(order_id = 1043)
Observation 1: status = shipped, courier = TCS, tracking = TX88

Thought 2: Ship ho gaya hai. Delivery ka andaza tracking se milega.
Action 2:  track_shipment(tracking = TX88)
Observation 2: "Out for delivery — Lahore hub, aaj shaam tak"

Thought 3: Ab mere paas poori maloomat hai. Jawab deta hoon.
Final Answer: Aap ka order TCS se nikal chuka hai aur aaj shaam tak pohanch jayega.
```

### ReAct kyun kaam karta hai?
| Tareeqa | Masla |
|---|---|
| Sirf Chain-of-Thought | Model apni yaadasht se "andaza" lagata hai → **hallucination** |
| Sirf Act (baghair soche tools) | Bina wajah tools chalata hai, tarteeb ghalat |
| **ReAct** | Soch ko **asli observations** se jor deta hai → grounded, aur trace parh kar debug bhi ho sakta hai |

### 🔄 2026 ki haqeeqat (interview mein extra points)
Aaj ke models mein **native tool calling** hai. "Thought / Action / Observation" ab text format mein likhna zaroori nahi:
- **Thought** → model ki (aksar chhupi) reasoning ya chhota sa text
- **Action** → structured tool call (JSON)
- **Observation** → tool result message

Yani **ReAct ab ek pattern hai, prompt format nahi.** Har modern "tool-calling loop" asal mein ReAct hi hai.

## 🗺️ Planning: Step-by-step vs Plan-first

### Step-by-step (ReAct style)
Har qadam par nayi soch. Plan dimagh mein chalta rehta hai.

### Plan-first (Plan-and-Execute)
```
  [Planner LLM] ──► Plan: 1) ... 2) ... 3) ...
        │
        ▼
  [Executor] step 1 → step 2 → step 3   (har step par tools)
        │
        ▼
  [Re-planner] nayi maloomat aayi? plan badlo ya jari rakho
```

| | Step-by-step (ReAct) | Plan-first |
|---|---|---|
| Lachak (flexibility) | Zyada — har qadam par adapt | Kam — plan par chalta hai (re-plan se behtar) |
| Lambe kaam mein coherence | Kam — bhatak sakta hai | Zyada — goal saamne rehta hai |
| Cost | Har step par bara model | Planner bara model, executor chhota/sasta ho sakta hai |
| User ko dikhana | Mushkil | Aasan — "yeh plan hai, approve karein?" |
| Best for | Chhote, unpredictable kaam | Lambe, multi-step kaam (research, migrations) |

**Plan Decomposition:** Bara goal → subtasks (aksar ek dependency order ke saath).
**Plan Revision:** Step 2 ka nateeja ghair mutawaqqo aaya → plan update. *Plan pathar ki lakeer nahi.*

Doosre naam jo interview mein aa sakte hain (sirf pehchan kafi hai):
- **ReWOO** — pehle poora plan placeholders ke saath, phir tools ek saath chala do (kam LLM calls).
- **Tree of Thoughts** — kayi soch ke raaste try kar ke behtareen chunna (mehnga).

## 🪞 Reflection & Self-Correction

**Reflection:** Agent apna output ya action dekh kar poochta hai *"kya yeh sahi hai? kya behtar ho sakta hai?"* aur sudharta hai.

- **Self-critique:** Same model apne jawab par tanqeed kare, phir dobara likhe.
- **Evaluator–Optimizer:** Alag "judge" prompt/model feedback de.
- **Reflexion (paper, 2023):** Fail hone par agent likhta hai *"maine kya ghalti ki"* aur yeh sabaq agli koshish mein memory ki tarah istemal karta hai.

> ⚠️ **Aham nuance (senior-level point):** Model aksar **sirf apne aap se** apni ghalti nahi pakar pata. Reflection tab sab se zyada kaam karti hai jab **bahar ka signal** ho — test fail hua, validator ne error diya, tool ne "not found" kaha. *"Dobara socho"* bina naye saboot ke aksar wahi ghalti dohrata hai, ya sahi jawab ko bhi ghalat bana deta hai.

## 🛑 Stopping Conditions & Loop Protection

Agent ko **kab rukna hai** — yeh design ka sab se nazar-andaaz hissa hai:

| Shart | Matlab |
|---|---|
| **Goal reached** | Model ne final answer diya / koi tool call nahi ki |
| **Max iterations** | Jaise 10–15 steps se zyada nahi |
| **Budget** | Token, cost ya waqt ki had |
| **No-progress detection** | Same tool + same arguments 2–3 bar → loop hai, roko |
| **Error threshold** | Lagataar failures → ruk kar insaan ko |
| **Human stop** | User ne cancel kiya |

Rukne par **gracefully** rukna: user ko batana kya ho gaya, kya baqi hai — khamoshi se gayab nahi hona.

## 💸 Cost of Planning
Zyada planning + reflection = behtar quality, magar **har extra LLM call = latency + paisa**. Chhote kaam par plan + reflect lagana aisa hai jaise chai banane ke liye project plan likhna.

## 🏭 Real-world
- **Support bot:** ReAct kaafi hai — 2–4 tool calls, plan ki zaroorat nahi.
- **Research agent:** Plan-first — pehle sub-questions, phir har ek par search, phir report.
- **Coding agent:** ReAct + **reflection with tests** — code likho, tests chalao, fail par error parh kar theek karo. Yahan reflection sab se zyada faida deti hai kyunke **test ek bahar ka sach** hai.

## 🎤 Interview Q&A — 5.2

**Q1. ReAct explain karein.**
> Reasoning aur acting ko interleave karna. Model sochta hai kya karna hai, tool chalata hai, observation dekhta hai, phir dobara sochta hai. Faida: reasoning asli data par grounded hoti hai, hallucination kam, aur trace se debugging aasan. Aaj native tool calling isi pattern ko structured shakal mein implement karti hai.

**Q2. Plan-and-Execute kab ReAct se behtar hai?**
> Lambe, multi-step kaam mein jahan coherence chahiye, user ko plan dikhana/approve karwana ho, ya cost bachani ho (bara model plan kare, chhota execute). ReAct chhote aur unpredictable kaamon mein behtar.

**Q3. Reflection hamesha faida deti hai?**
> Nahi. Bahar ke feedback (tests, validators, tool errors) ke saath kaafi faida; sirf "khud check karo" se aksar faida kam aur cost/latency zyada. Main reflection wahan lagaunga jahan verify karne ka objective tareeqa ho.

**Q4. Agent infinite loop mein phans gaya. Kya karenge?**
> Fori: max-iterations aur budget cap. Phir wajah: aksar tool ka ghair-wazeh error message (model ko pata hi nahi kya badle), ya ek jaise do tools. Repeated same-call detection lagaunga, error messages ko "agla qadam kya ho" batane wala banaunga, aur trace se dekhunga loop kahan shuru hua.

**Q5. Agent ke rukne ki shartain kya honi chahiye?**
> Goal reached, max steps, token/cost/time budget, no-progress detection, error threshold aur user cancel — aur rukne par user ko saaf status.

## ⚠️ Aam galtiyan
- Stop conditions sirf "max steps" — no-progress detection bhool jana
- Har kaam mein reflection/planning laga dena
- Plan ko ek bar bana kar kabhi revise na karna

## ✅ Yaad rakhne wali baat
> **ReAct = soch ko saboot se jorna.** Planning lambe kaam ke liye, reflection tab jab bahar ka feedback ho, aur **rukne ki shart har agent ki zaroorat.**

# 5.3 Tools for Agents — Agent ke haath paon

🎯 **Ek line mein:** Agent utna hi acha hai jitne uske tools; tool ki **description bhi ek prompt hai**, aur har tool ek **security darwaza** hai.

## 🧠 Tool asal mein kya hai?
Tool ek **contract** hai: `naam + description + input schema + output`.
- Model sirf **request** karta hai ("is tool ko yeh arguments do").
- Tool **aap ki application chalati hai**, model nahi.
- Nateeja model ko **observation** ke tor par wapas jata hai.

**Agentic tool use** (L2 se farq): L2 mein aksar ek tool call; yahan model **bar bar, apni marzi se** dastiyab tools mein se chunta hai, aur ek tool ka nateeja agle tool ke arguments tay karta hai.

## 🛠️ Agent-Computer Interface (ACI)
Anthropic ka mashwara: jitni mehnat log **user interface (UI)** par karte hain, utni hi **agent interface (tools)** par karo. Model ko apna tool aise samjhao jaise naye developer ko function ka documentation dete ho.

### Achhe tool design ke usool

| Usool | Bura | Acha |
|---|---|---|
| **Wazeh naam** | `search` | `get_order_by_id`, `search_products_by_name` |
| **Description mein "kab" aur "kab nahi"** | "Gets order" | "Customer ke order ka status lao. Sirf tab jab order number maloom ho; warna pehle customer se poocho." |
| **Kam aur typed parameters** | `query: string` (sab kuch isi mein) | `order_id: integer`, `status: enum[pending, shipped, delivered]` |
| **Ghalti mushkil banao (poka-yoke)** | Relative date "kal" | Wazeh format "YYYY-MM-DD" maango |
| **Kaam ka output** | Raw UUIDs, 200 fields | Insaani naam + sirf zaroori fields |
| **Error jo raasta dikhaye** | `Error 404` | "Order nahi mila. Customer se order number confirm karein (format: #1001–#9999)." |

> 💡 **Error message model ke liye agla instruction hai.** Ghair-wazeh error = agent ka loop mein phansna.

## 📏 Tool granularity: kitna bara tool?
- **Bohot chhote tools** (list_users, list_events, create_event alag) → bohot se calls, zyada ghaltiyon ka mauqa.
- **Bohot bara tool** (`do_everything`) → model ko samajh nahi aata kab kya hoga.
- **Behtar:** asli kaam ke mutabiq tools — `schedule_meeting(attendees, time)` jo andar 3 kaam khud kare.

## 🔢 Kitne tools?
Jitne tools zyada aur milte julte, utna model ka **ghalat tool chunna** barhta hai (aur descriptions context window bhi khati hain). Hal:
- Tools ko **kam aur alag alag** rakho (overlap khatam)
- **Dynamic tool exposure:** sirf is kaam ke mutalliqa tools dikhao (tool search / grouping)
- Bohot zyada domains → **alag specialist agents** (5.6)

## 🔗 Dependent, Parallel aur Repeated calls
- **Dependent:** `find_customer(phone)` → customer_id → `get_orders(customer_id)` → order_id → `track_shipment(...)`
- **Parallel:** "Yeh 3 ghariyan stock mein hain?" → teen `check_stock` ek saath.
- **Repeated:** ek hi tool alag arguments ke saath kayi bar.

## 📦 Tool result ka size mehdood rakhna
1000 orders ka JSON wapas dena = context window bhar gayi, model ka dhyan bikhar gaya, paisa zaya.
- **Filtering / pagination:** `limit=10`, "aur chahiye to page 2"
- **Truncation with note:** "50 mein se pehle 10 dikhaye; filter use karein"
- **Summary + ID:** khulasa do, detail chahiye to ID se dobara fetch
- **Sirf zaroori fields** (naam, status, tareekh — 40 columns nahi)

## 🔐 Read-only vs Consequential Tools — Permission Tiers

| Tier | Qism | Watch store example | Policy |
|---|---|---|---|
| 0 | **Read-only** | `get_order_status`, `check_stock`, `get_return_policy` | Khud chalao |
| 1 | **Reversible write** | `create_exchange_request`, `add_note_to_order` | Khud chalao + log |
| 2 | **Irreversible / paisa / bahar bhejna** | `issue_refund`, `cancel_order`, `send_email_to_all` | **Human approval** |
| 3 | **Mana** | DB delete, doosre customer ka data | Agent ko tool hi mat do |

> 🔑 Sab se mazboot security: **jo tool nahi diya, woh misuse nahi ho sakta.** (Least privilege)

## 🛡️ Arguments ko untrusted samjho
Model ke diye arguments ko **code mein validate** karo:
- Type aur range check (refund amount order amount se zyada to nahi?)
- **Authorization:** kya yeh customer **sirf apna** order dekh raha hai? Agent ko "customer_id" model se nahi, **session se** do — warna prompt injection se koi doosre ka order mangwa lega (*confused deputy* masla).

## 🔧 Tool failure recovery
| Failure | Kya karein |
|---|---|
| Timeout / 503 (temporary) | Code mein retry with backoff |
| Ghalat arguments | Saaf error **model ko observation** ki tarah wapas — woh theek kare |
| Tool hi down | Fallback tool ya graceful message ("abhi system slow hai, 10 min mein dobara") |
| Bar bar fail | Ruk kar insaan ko escalate |

> Tool error par **poora agent crash mat karo** — error ko observation bana kar model ko do.

## 🎤 Interview Q&A — 5.3

**Q1. Agent ke liye acha tool kaise design karenge?**
> Wazeh naam, aisi description jo bataye kab use karna hai aur kab nahi, kam aur strongly-typed parameters (enums), sirf zaroori output, aur aise error messages jo model ko agla qadam batayein. Tools overlap na karein. Main tool descriptions ko prompt ki tarah test aur iterate karta hoon.

**Q2. Agent ke paas 50 tools hain aur ghalat tool chun raha hai. Kya karenge?**
> Pehle overlap khatam aur milte julte tools merge. Phir dynamic exposure — har step par sirf mutalliqa tools. Agar domains bilkul alag hain to specialist sub-agents jinhein kam tools milein. Aur tool selection ko eval set se naapunga (L7).

**Q3. Kya agent ko delete ya refund ka access dena chahiye?**
> Consequential actions ko tier karna chahiye. Irreversible ya paise wale actions par human approval, code-level validation (amount limits), idempotency keys, aur audit log. Bilkul khatarnak actions ka tool hi nahi dena.

**Q4. Tool ka output bohot bara hai, kya karenge?**
> Filtering, pagination, sirf zaroori fields, truncation ke saath note, ya poora nateeja bahar store kar ke model ko khulasa + ID dena.

**Q5. Prompt injection se agent doosre user ka data kaise nikal sakta hai, aur kaise rokenge?**
> Agar user identity model ke argument se aati hai to injection se badli ja sakti hai. Isliye identity aur permissions session/backend se aati hain, aur har tool call par code mein authorization check hota hai — model par bharosa nahi.

## ⚠️ Aam galtiyan
- Tool description ek lafz ki ("Gets data")
- Raw API ke 30 endpoints ko seedha 30 tools bana dena
- Error par sirf status code wapas dena
- Validation sirf prompt mein ("refund 5000 se zyada mat dena") — code mein nahi

## ✅ Yaad rakhne wali baat
> **Tool description = prompt. Error message = agla instruction. Har tool = security darwaza.**

# 5.4 MCP (Model Context Protocol) — Tools jorne ka "USB-C"

🎯 **Ek line mein:** MCP ek **open standard** hai jis se koi bhi AI app (host) kisi bhi tool/data source (server) se **ek hi tareeqe** se jur sakti hai.

## 🧠 Masla jo MCP hal karta hai: N × M

MCP se pehle: 5 AI apps × 10 tools = **50 alag integrations** — har app har tool ke liye apna connector likhti.

MCP ke baad: har tool ek **MCP server** banaye, har app ek **MCP client** rakhe → **5 + 10 = 15**. Ek bar banao, har jagah chalao.

> 🔌 **USB-C misaal:** Pehle har phone ka alag charger; ab ek port, sab devices. MCP AI ke liye wohi port hai.

MCP **Anthropic ne Nov 2024** mein introduce kiya; ab OpenAI, Google, Microsoft aur bohot se tools ne bhi apnaya hai — is liye yeh industry standard ban chuka hai.

## 🏗️ Architecture: Host → Client → Server

```
 ┌───────────────────────── HOST (AI app: Claude Desktop, IDE, aap ka agent) ──┐
 │                                                                             │
 │   ┌──────────┐        ┌──────────┐        ┌──────────┐                       │
 │   │ Client 1 │        │ Client 2 │        │ Client 3 │   (har server ka      │
 │   └────┬─────┘        └────┬─────┘        └────┬─────┘    apna 1 client)     │
 └────────┼───────────────────┼───────────────────┼─────────────────────────────┘
          │ stdio             │ HTTP              │ HTTP
          ▼                   ▼                   ▼
   ┌─────────────┐     ┌─────────────┐     ┌─────────────┐
   │ MCP Server  │     │ MCP Server  │     │ MCP Server  │
   │ Filesystem  │     │  Shopify    │     │   GitHub    │
   └─────────────┘     └─────────────┘     └─────────────┘
```

| Hissa | Kaam |
|---|---|
| **Host** | AI application jahan user baat karta hai; LLM aur permissions yahin |
| **Client** | Host ke andar connector; **ek server se 1:1 connection** |
| **Server** | Program jo tools/data expose karta hai (local ya remote) |

Messages **JSON-RPC 2.0** format mein hote hain. Connection shuru hote hi dono **capabilities negotiate** karte hain (server kya kya deta hai, client kya support karta hai).

## 🧰 MCP ke Primitives — kaun kya deta hai

### Server ki taraf se (3 cheezen)
| Primitive | Kya hai | Kaun control karta hai | Example |
|---|---|---|---|
| **Tools** | Actions jo model call kar sake | **Model** decide karta hai | `create_issue`, `get_order` |
| **Resources** | Data/context jo parha ja sake (file, DB record) | **Application** decide karti hai kya attach karna | Ek file ka content, product catalog |
| **Prompts** | Dobara istemal hone wale templates/workflows | **User** chunta hai (jaise slash command) | "/summarize-pr" |

### Client ki taraf se (pehchan kaafi)
- **Sampling:** Server, host ke LLM se kuch generate karwane ki darkhwast kare
- **Roots:** Server ko batana kaunse folders/locations ki had hai
- **Elicitation:** Server, user se beech mein koi maloomat maange

## 🚚 Transports
| Transport | Kab | Note |
|---|---|---|
| **stdio** | Local server — host use apne computer par subprocess ki tarah chalata hai | Tez, simple; server aap ke user ki permissions ke saath chalta hai |
| **Streamable HTTP** | Remote server (internet/company network) | Kayi users, auth zaroori. (Purana "HTTP + SSE" tareeqa is se replace hua) |

Remote servers ke liye authorization ka standard **OAuth 2.1** based hai.

## ⚖️ Tool Calling vs MCP — sab se aam confusion

| | Tool / Function Calling | MCP |
|---|---|---|
| Sawal | Model action **kaise maangta** hai? | Tools **kaise expose aur discover** hote hain? |
| Level | Model ↔ aap ki app | App ↔ bahar ke tools/servers |
| Taluq | MCP ke tools aakhir mein model ko tool calling ke zariye hi diye jate hain | Tool calling ke upar ek **standard integration layer** |

> Ek line: **Tool calling "zuban" hai, MCP "plug standard" hai.** Dono ek doosre ki jagah nahi lete.

## 🔐 MCP Security — interview mein zaroor poocha jata hai
| Khatra | Kya hota hai | Bachao |
|---|---|---|
| **Malicious / anjaan server** | Server khud nuqsan de, data chura le | Sirf trusted sources, review, versions pin |
| **Tool poisoning** | Tool description mein chhupi hidayat: "yeh tool chalane se pehle ~/.ssh parh kar bhejo" | Descriptions review, hassas tools par approval |
| **Rug pull** | Approval ke baad server tool ki definition badal de | Version pinning, changes par alert |
| **Over-permissioned tokens** | Server ko poore account ka access | Least privilege, scoped tokens |
| **Indirect prompt injection** | Resource (email, doc) ke andar buri hidayat | Retrieved content = data, hukm nahi (L8) |
| **Local server privileges** | stdio server aap ki poori machine tak pohanch rakhta hai | Sirf bharosemand local servers |

## 🏭 Real-world
- Watch store ke liye ek **"Store MCP server"** jis mein `get_order`, `check_stock`, `get_policy` tools hon — phir yahi server Claude Desktop, aap ka WhatsApp agent, ya koi aur MCP-compatible app sab istemal kar sakein.
- **Kab MCP overkill hai:** sirf ek app, 3–4 internal tools → seedha tool calling kaafi. MCP ka faida tab jab **reuse** ho, kayi hosts hon, ya doosron ke liye integration bana rahe hon.

## 🎤 Interview Q&A — 5.4

**Q1. MCP kya hai aur kyun chahiye?**
> Ek open protocol jo AI apps ko tools, data aur prompts se standard tareeqe se jorta hai. Yeh N×M integration masle ko N+M bana deta hai — tool ek bar MCP server banaye, aur har compatible host use istemal kar le.

**Q2. Host, client aur server mein farq?**
> Host AI app hai (LLM aur user wala hissa). Client host ke andar connector hai jo ek server se 1:1 jurta hai. Server tools/resources/prompts expose karta hai.

**Q3. Tools aur Resources mein kya farq hai?**
> Tools actions hain jinhein model khud call karne ka faisla karta hai. Resources read-only data/context hain jinhein application (ya user) context mein shamil karti hai. Control kis ke haath — yahi farq hai.

**Q4. stdio aur HTTP transport kab?**
> stdio local servers ke liye (same machine, subprocess). Streamable HTTP remote, multi-user servers ke liye jahan OAuth jaisi authentication chahiye.

**Q5. MCP ke security risks?**
> Malicious servers, tool poisoning (descriptions mein chhupi hidayat), rug pulls, zyada permissions wale tokens, aur resources ke zariye indirect prompt injection. Bachao: trusted servers, version pinning, least privilege, description review aur consequential actions par human approval.

**Q6. Function calling ke hote hue MCP kyun?**
> Function calling batata hai model tool kaise maange; MCP standardize karta hai tools kaise discover, connect aur reuse hon — kayi apps aur teams ke beech.

## ⚠️ Aam galtiyan
- MCP ko "function calling ka replacement" samajhna
- Internet se koi bhi MCP server baghair dekhe install kar lena
- Har chhote project mein MCP laga dena jahan reuse hai hi nahi

## ✅ Yaad rakhne wali baat
> **MCP = AI ka USB-C.** Host → Client → Server; Tools (model), Resources (app), Prompts (user). Aur **har server ek trust decision hai.**

# 5.5 Memory Basics — Agent ko yaad rakhna sikhana

🎯 **Ek line mein:** LLM API **stateless** hai — jo yaad rakhna hai woh aap ka system rakhta hai; asal hunar yeh hai ke **kya** yaad rakhein aur **kab** wapas laayein.

## 🧠 Teen cheezen jo log mix kar dete hain

| | Kya hai | Misaal |
|---|---|---|
| **History** | Pichle messages ka raw record | Poori WhatsApp chat |
| **State** | Abhi kaam kahan khara hai | "Exchange request bana di, approval ka intezar" |
| **Memory** | Jo **yaad rakhne layak** hai (aksar chhaant kar) | "Customer Lahore mein rehta hai, black strap pasand karta hai" |

## 💻 RAM aur Hard Disk wali misaal
- **Context window = RAM:** tez, magar mehdood aur har request ke baad saaf.
- **Database / vector store = Hard disk:** bari, mustaqil, magar wahan se **sahi cheez dhoond kar** laani parti hai.
- Memory system ka kaam: **disk se sahi cheez RAM mein sahi waqt par laana.**

## ⏱️ Short-term vs Long-term

| | Short-term memory | Long-term memory |
|---|---|---|
| Scope | Ek conversation / ek task | Conversations ke darmiyan |
| Kahan | Context window + state (thread) | Database / vector store |
| Masla | Window bhar jaati hai | Sahi cheez retrieve karna, purani cheez update karna |
| LangGraph mein | **Checkpointer** (per `thread_id`) | **Store** (cross-thread, namespaces) |

## 🗂️ Long-term memory ki qismein (CoALA framework — pehchan kaafi)
| Qism | Kya yaad | Example |
|---|---|---|
| **Semantic** | Facts | "Customer ka naam Ali, shehar Lahore" |
| **Episodic** | Pichle tajurbe / waqiat | "Pichli bar exchange mein courier late hua tha, customer naraz tha" |
| **Procedural** | Kaam kaise karna hai (rules/instructions) | "Is customer ko hamesha Urdu mein jawab" / system prompt ki updates |

## 🔄 Memory ke chaar bunyadi kaam: Write → Retrieve → Update → Forget

**1. Write (kya save karein?)**
- ✅ Durable facts, preferences, faisle, task ke nateeje, summaries
- ❌ Har message, small talk, wahi cheez jo dobara API se mil sakti hai (order status — yeh live data hai, memory nahi!)

*Kab likhein?*
| Hot path | Background |
|---|---|
| Agent conversation ke dauran khud "save_memory" tool chalaye | Conversation ke baad ek alag process chat parh kar memory nikale |
| Foran dastiyab, magar jawab slow aur agent ka dhyan bata | Jawab tez, memory thori der se, behtar chhaant |

**2. Retrieve (kya wapas laayein?)** — har bar sab kuch nahi; sirf mutalliqa. Retrieval aksar teen cheezon ka mel: **relevance** (similarity), **recency** (taazgi), **importance** (ahmiyat).

**3. Update** — facts badalte hain: "pehle Karachi, ab Lahore". Purani yaad ko update karo, warna takrao.

**4. Forget** — purani/ghalat yaadein hatao; user kahe "bhool jao" to sach mein mitao; privacy qawaneen ke mutabiq.

## ✂️ Barhti conversation sambhalna (Trimming & Summarizing)
| Tareeqa | Kaise | Nuqsan |
|---|---|---|
| **Last-N messages** | Sirf aakhri 10 messages | Shuru ki aham baat (order #) kho sakti hai |
| **Token-based trim** | Had tak messages rakho | Wahi masla |
| **Summarize old turns** | Purane hisse ki summary + naye messages poore | Summary mein numbers/IDs kho sakte hain |
| **Hybrid** (behtareen) | System prompt + **running summary** + **pinned facts** (order #, naam) + aakhri messages | Thora zyada kaam |

> 🔑 **Exact data (IDs, amounts, dates) ko summary par mat chhoro** — unhein alag "pinned state" mein rakho. (Yeh L6 Context Engineering ka bhi bunyadi sabaq hai.)

## 🛡️ Memory ke khatre
- **Stale memory:** purani baat ko sach maan lena
- **Memory poisoning:** user/document ki buri hidayat memory mein save ho kar har conversation mein chalti rahe
- **Cross-user leakage:** ek customer ki memory doosre ko → **har memory ko user ke namespace** mein rakho, retrieval hamesha user_id se filter
- **Privacy:** PII, hassas maloomat ko save na karna, ya mehfooz rakhna; delete ka raasta

## 🏭 Real-world (watch store bot)
- **Short-term:** is chat ka state — kaunsa order discuss ho raha, exchange kis stage par.
- **Long-term:** customer ki preference (Urdu/English), pichle orders ka khulasa, "pichli bar delivery late hui thi".
- **Memory nahi:** live order status, stock — yeh har bar **tool** se taaza lao.

## 🎤 Interview Q&A — 5.5

**Q1. Short-term aur long-term memory mein farq?**
> Short-term ek conversation/task ke andar hai — context window aur thread state. Long-term conversations ke darmiyan persist hoti hai — DB ya vector store se retrieve hoti hai. LangGraph mein checkpointer short-term, store long-term.

**Q2. Lambi chat mein agent customer ka order number bhool jata hai. Fix?**
> Context truncation ya summary mein ID kho rahi hai. Aham entities (order #, naam) ko alag pinned state mein rakhunga jo har request mein jaye, purani history ki running summary banaunga aur aakhri messages poore rakhunga.

**Q3. Memory mein kya store karna chahiye aur kya nahi?**
> Durable facts, preferences, faisle aur summaries. Live data (jo tool se mil sakta hai), small talk aur hassas data nahi. Har memory durability ke test par: kya mahine baad bhi sach aur kaam ka hoga?

**Q4. Multi-user system mein memory leak kaise rokenge?**
> Har memory user/tenant namespace mein; retrieval har bar user_id filter ke saath, aur user_id session se aata hai, model se nahi.

**Q5. Hot-path vs background memory writing?**
> Hot path foran memory banata hai magar latency barhata aur agent ka dhyan baant'ta hai. Background process jawab ko tez rakhta aur behtar chhaant karta hai, magar memory thori der se available hoti hai.

## ⚠️ Aam galtiyan
- Poori history har bar bhejna ("memory" samajh kar)
- Live data ko memory bana dena (stale ho jata hai)
- Update/forget ka nizaam na banana
- Sab users ki memory ek hi collection mein bina filter

## ✅ Yaad rakhne wali baat
> **History ≠ State ≠ Memory.** Write → Retrieve → Update → Forget. Memory yaad ke liye hai; live data tools se aata hai.

# 5.6 Multi-Agent Architectures — Kayi agents ki team

🎯 **Ek line mein:** Multi-agent tabhi lagao jab **specialization ka faida, extra complexity aur cost se zyada ho** — warna ek mazboot agent behtar.

## 🧠 Multiple agents kyun?
- **Alag tools/permissions:** billing agent ko refund tool, FAQ agent ko nahi
- **Alag maharat / hidayat:** har agent ka focused system prompt
- **Alag models:** sasta model aasan kaam par, bara model mushkil par
- **Parallel kaam:** 5 sub-topics ek saath research
- **Context ki safai:** har sub-agent ko saaf, chhota context — lead ko sirf condensed nateeja

## 🏗️ Bare patterns

### 1. Orchestrator–Workers / Supervisor
```
                 ┌──────────────┐
   User ───────► │  Supervisor  │ ◄──── nateeje jorta hai
                 └──┬────┬────┬─┘
                    ▼    ▼    ▼
              [Orders][Products][Billing]   ← workers / specialists
```
Supervisor kaam baant'ta, route karta aur jorta hai. **Sab se aam aur controllable.**

### 2. Sequential Handoffs (Relay race)
`Intake agent → Diagnosis agent → Resolution agent` — har ek kaam khatam kar ke agle ko **control + context** deta hai. (OpenAI Agents SDK mein "handoffs" isi ka naam hai.)

### 3. Parallel Specialists
Ek hi waqt mein alag pehlu: legal reviewer, technical reviewer, tone reviewer → phir merge.

### 4. Hierarchical Supervisor
Supervisor ke neeche supervisors, unke neeche workers — bari organization jaisa. Bohot bare systems ke liye; complexity zyada.

### 5. Network / Swarm (pehchan kaafi)
Koi markazi boss nahi; agents aapas mein ek doosre ko kaam de dete hain. Flexible magar control aur debugging mushkil.

## 💥 Kya ghalat hota hai (What Goes Wrong)
| Masla | Wazahat |
|---|---|
| **Context ka kho jana** | Sub-agent ko poora pas-e-manzar nahi mila, ghalat samjha |
| **Dohra kaam** | Do agents ne ek hi search kar di |
| **Takrate nateeje** | Ek ne "exchange", doosre ne "refund" kaha |
| **Coordination overhead** | Agents ke darmiyan messages hi tokens kha gaye |
| **Latency & cost** | Har agent = alag LLM calls |
| **Debugging** | Ghalti kis agent ki? Trace mushkil |

## 📊 Industry ki do aawazein (senior-level context)
- **Anthropic (multi-agent research system):** Lead agent kayi subagents ko **parallel** mein research par bhejta hai — **breadth-first, parallelizable** kaam mein quality kaafi behtar hui. Magar token kharcha bohot zyada: simple chat ke muqable mein takreeban **15 guna** tokens. Faida sirf **qeemti kaamon** mein.
- **Cognition (Devin) ka "Don't build multi-agents" argument:** Agents ke darmiyan **context share** na ho to woh ek doosre ke faislon se be-khabar ho kar takrate faisle karte hain; **ek agent jis ke paas poora context ho** aksar zyada reliable.

> 🎯 Dono baatein sach hain: **parallel, alag alag research** jaise kaam multi-agent ke liye ache; **gehre jure hue faisle** (jaise code likhna) ek agent ke liye behtar.

## 🤝 A2A vs MCP

| | **MCP** | **A2A (Agent-to-Agent)** |
|---|---|---|
| Jorta hai | Agent ↔ **tools / data** | Agent ↔ **doosra agent** |
| Rukh | "Vertical" — neeche tools ki taraf | "Horizontal" — barabar ke agents |
| Misaal | Agent → Shopify server | Aap ka support agent → courier company ka agent |
| Kis ne | Anthropic (2024) | Google (2025) |
| Khaas cheez | Tools/Resources/Prompts | **Agent Card** (agent apni salahiyatein batata hai), tasks, alag organizations ke "opaque" agents |

> Ek line: **MCP agent ko haath deta hai, A2A agent ko saathi deta hai.** Dono mukhalif nahi, ek doosre ke saath chalte hain.

## ⚖️ Kab ek agent kaafi hai?
Pehle **single agent** se shuru karo. Multi-agent mein tab jao jab:
1. Tools itne zyada ho gaye ke tool selection gir rahi hai
2. Alag hisson ko **alag permissions** chahiye (security boundary)
3. Kaam asal mein **parallel** ho sakta hai
4. Context window ek agent ke liye kaafi nahi rahi
5. Aur — **evals se saabit** ho ke multi-agent behtar hai

## 🏭 Real-world (watch store)
Shuru mein: **ek agent** + 6–8 tools. Baad mein agar business barhe:
- **Supervisor/router** → `Orders agent` (tracking, exchange), `Sales agent` (recommendations, stock), `Escalation` (insaan ko handoff)
- Har agent ke tools aur permissions alag — sales agent refund nahi kar sakta.

## 🎤 Interview Q&A — 5.6

**Q1. Multi-agent system kab banayenge?**
> Jab specialization ka asli faida ho — alag tools/permissions, parallel kaam, ya context ki had. Default single agent; multi-agent tab jab evals dikhayein ke yeh behtar hai, kyunke yeh cost, latency aur coordination ki pechidgi barhata hai.

**Q2. Supervisor aur swarm/handoff pattern mein farq?**
> Supervisor mein ek markazi agent route aur control karta hai — predictable aur debug karna aasan. Handoff/swarm mein agents control ek doosre ko de dete hain — flexible magar flow trace karna mushkil.

**Q3. Multi-agent ke bare masle?**
> Context loss, dohra kaam, takrate faisle, coordination overhead, cost (research systems mein kai guna tokens), aur debugging. Hal: wazeh roles, shared state, structured handoffs, aur har agent ke output ka contract.

**Q4. MCP aur A2A mein farq?**
> MCP agent ko tools/data se jorta hai; A2A agents ko doosre agents se, aksar alag systems ya organizations ke darmiyan, Agent Cards ke zariye discovery ke saath.

**Q5. Agents ke darmiyan context kaise pass karenge?**
> Shared state object (jaise LangGraph state), structured handoff message (goal, ab tak kya hua, constraints, expected output), aur sub-agent se sirf condensed nateeja wapas — poori transcript nahi.

## ⚠️ Aam galtiyan
- "Multi-agent cool lagta hai" is liye bana dena
- Sub-agent ko sirf ek line ka kaam dena, pas-e-manzar nahi
- Agents ke output ka koi fixed format na hona

## ✅ Yaad rakhne wali baat
> **Ek mazboot agent se shuru karo.** Multi-agent parallel aur specialized kaam ke liye; har agent = extra cost aur coordination.

# 5.7 Building Agents with LangGraph — Concepts (baghair code)

🎯 **Ek line mein:** LangGraph agent ko ek **graph** ki tarah banata hai — nodes (kaam), edges (raaste), aur ek shared **state** — taake control flow wazeh, resumable aur debuggable ho.

## 🧠 Kyun LangGraph?
Seedha "agent loop" library mein chhupa hota hai — aap ko pata nahi andar kya ho raha. LangGraph kehta hai: **flow ko khud draw karo.**
- Kab model chale, kab tool, kab insaan ki approval — sab **explicit**
- Har step ke baad state **save** — crash ho to wahin se shuru
- Loops, branches, human-in-the-loop built-in

## 🧱 Bunyadi building blocks

| Concept | Kya hai | Misaal |
|---|---|---|
| **State** | Graph mein safar karta shared data (schema) | messages, order_id, plan, approval_status |
| **Reducer** | State ka field kaise update ho — replace ya **append** | messages hamesha append; status replace |
| **Node** | Kaam ki ek unit (model call, tool, logic) | `classify`, `call_model`, `run_tools` |
| **Edge** | Ek node se doosre ka raasta | `classify → faq` |
| **Conditional Edge** | State dekh kar agla node chunna | "tool call hai? → tools, warna → END" |
| **START / END** | Shuru aur khatam | |
| **Cycle / Loop** | Model ⇄ Tools ka dairah | ReAct loop |
| **Checkpointer** | Har step par state save (thread_id ke saath) | Crash/pause ke baad resume |
| **Interrupt** | Graph ko rok kar insaan ka input | Refund approval |
| **Subgraph** | Graph ke andar graph | Har specialist agent apna subgraph |

## 🗺️ Watch store support agent ka graph

```
  START
    │
    ▼
 [classify] ──── FAQ sawal ──────────────► [faq_answer] ──► END
    │
    │ order / complex
    ▼
 [agent (LLM)] ◄──────────────┐
    │                         │
    ├── tool call hai? ──► [tools] ─┘   (loop)
    │
    ├── refund / cancel? ──► [human_approval]  ⏸ INTERRUPT
    │                               │
    │                     approve ──┴──► [execute_action] ──► [agent]
    │                     reject  ─────► [agent] (customer ko batao)
    │
    └── final answer ──► END
```

## 💾 Checkpointing & Persistence
- Har node ke baad state ek **checkpoint** ki shakal mein save (memory, SQLite, Postgres waghera).
- Har conversation ka ek **thread_id** — wahi short-term memory hai.
- Faide:
  - **Resume:** server restart, crash ya lamba intezar — wahin se jari
  - **Human-in-the-loop:** ghanton baad approval aaye to bhi graph wahin se chale
  - **Time travel:** purane checkpoint par ja kar dekhna ya wahan se naya raasta (fork) — debugging ke liye kamaal
  - **Fault tolerance:** ek node fail ho to pichla kaam zaya nahi

## 🙋 Human-in-the-Loop (HITL) Interrupts
Graph ek node par **ruk jata** hai, insaan ko dikhata hai kya hone wala hai, aur insaan:
- ✅ **Approve** — waisa hi chalao
- ✏️ **Edit** — arguments badlo (refund 3000 nahi, 2000)
- ❌ **Reject** — mat karo, wajah ke saath agent ko wapas

Yeh checkpointing ke baghair mumkin nahi — kyunke rukne ke dauran state kahin mehfooz honi chahiye.

## 🔍 State Inspection & Streaming
- **Inspection:** kisi bhi waqt dekh sakte ho state mein kya hai aur kaunsa raasta kyun liya — "black box" nahi
- **Streaming:** sirf aakhri jawab nahi — har node ka update, ya LLM ke tokens live user ko dikhana (progress UX, 5.9)

## 🧭 "Prebuilt ReAct agent" vs custom graph
- LangGraph mein ek **ready-made ReAct agent** bhi hai (model + tools ka loop) — jaldi shuru ke liye.
- Custom graph tab jab: classification, approval steps, alag branches, ya multi-agent chahiye.

## 🎤 Interview Q&A — 5.7

**Q1. LangGraph kyun, seedha agent loop kyun nahi?**
> Explicit control flow, built-in persistence/checkpointing, human-in-the-loop interrupts, loops aur branches, aur state inspection. Production mein mujhe pata hona chahiye agent ne kaunsa raasta kyun liya aur crash ke baad resume ho sake.

**Q2. State aur reducer kya hain?**
> State graph ka shared data schema hai jo har node parhta/update karta hai. Reducer batata hai field kaise update ho — messages ko append, status ko replace.

**Q3. Checkpointing ka faida?**
> Har step ke baad state save hoti hai thread_id ke saath — resume, human approval ka intezar, time-travel debugging aur fault tolerance mumkin hote hain.

**Q4. Human-in-the-loop kaise implement karenge (conceptually)?**
> Consequential action se pehle interrupt; state checkpoint mein save; insaan ko proposed action dikhao; approve/edit/reject par graph usi checkpoint se resume.

**Q5. Conditional edge ki misaal?**
> Model node ke baad: agar response mein tool call hai to tools node, refund hai to approval node, warna END.

## ⚠️ Aam galtiyan
- Har cheez ek bare node mein thoons dena (phir graph ka faida hi nahi)
- Checkpointer ke baghair HITL banane ki koshish
- State mein poore bare tool outputs rakhna (context aur storage dono bharte hain)

## ✅ Yaad rakhne wali baat
> **State + Nodes + Edges + Checkpointer.** LangGraph ka asal tohfa: **wazeh flow, resume, aur human-in-the-loop.**

# 5.8 Sandboxed Execution — Agent ke likhe code ko mehfooz chalana

🎯 **Ek line mein:** Model ka likha code **ajnabi ka code** hai — use hamesha ek alag, band, mehdood mahol (sandbox) mein chalao.

## 🧠 Kyun zaroori?
Coding/data agents code likhte aur chalate hain (data analysis, chart, file conversion). Agar yeh code **aap ke server ke andar** (in-process) chale to:
- `rm -rf` jaisi ghalti → files khatam
- Environment variables se **API keys parh kar** bahar bhej de
- Prompt injection se likhwaya gaya code → **data exfiltration**
- Infinite loop / bohot memory → server down
- Network se kuch bhi download / upload

> Model ki niyat buri nahi hoti — magar **ghalti, hallucination ya prompt injection** kaafi hai.

## 🧱 Isolation ki seerhi (kamzor → mazboot)

| Level | Tareeqa | Hifazat |
|---|---|---|
| ❌ | Same process mein `exec` | Koi nahi |
| 1 | Alag process | Thori |
| 2 | **Container (Docker)** | Achhi — magar host ka kernel share |
| 3 | gVisor jaisa user-space kernel | Zyada |
| 4 | **MicroVM (Firecracker)** | Bohot mazboot, phir bhi tez |
| 5 | Alag VM / alag machine | Sab se zyada, mehnga |

Services: **E2B** (agents ke liye cloud sandboxes, microVM based), **Modal** (cloud containers/functions), apne **Docker containers** — sab isi kaam ke liye.

## 🔒 Kya kya band karna hai

| Resource | Paabandi |
|---|---|
| **Filesystem** | Sirf ek working folder; host files nazar hi na aayein |
| **Network** | Default band; zaroorat ho to sirf allowlist domains |
| **Time** | Har run ka timeout (jaise 30 sec) |
| **CPU / Memory** | Had (fork bomb, memory bomb se bachao) |
| **Credentials** | Sandbox mein **koi secret nahi** |
| **Lifetime** | **Ephemeral** — har session naya sandbox, kaam ke baad khatam |

## 📤 Safe Output Capture
- `stdout` / `stderr` pakro — magar **truncate** karo (10 MB log model ko mat do)
- Bani files ko **artifacts** ki tarah bahar nikalo (chart image, CSV)
- Model ko **structured nateeja**: success/fail, output ka khulasa, error ki aakhri lines

## 💥 Blast Radius
> *"Agar sab se bura ho jaye to nuqsan kahan tak phailega?"*

Design ka maqsad: blast radius **chhota** — ek ephemeral sandbox toota, bas. Na database, na secrets, na doosre users.

## 🏭 Real-world: Code-interpreter loop
```
 Agent: "sales.csv ka monthly chart banao"
   → code likhta hai → [SANDBOX] chalata hai
   → error: column 'Date' nahi mila
   → error parh kar code theek → dobara chalata hai
   → chart.png bana → artifact user ko
```
Yeh **reflection with real feedback** ki behtareen misaal hai (5.2) — error ek asli, bahar ka signal hai.

## 🎤 Interview Q&A — 5.8

**Q1. Agent ka generate kiya code kaise chalayenge?**
> Hamesha isolated, ephemeral sandbox mein — container ya behtar microVM (E2B jaisi service). Filesystem scoped, network default off ya allowlist, timeout, CPU/memory limits, aur koi credentials nahi. Output truncate kar ke structured shakal mein model ko.

**Q2. Kya Docker container kaafi hai?**
> Bohot se cases mein theek, magar container host ka kernel share karta hai, is liye untrusted code ke liye gVisor ya microVM zyada mehfooz hai. Faisla risk aur multi-tenancy par: doosre users ka code bhi chal raha ho to mazboot isolation.

**Q3. Network kyun band karna chahiye?**
> Data exfiltration aur malicious downloads rokne ke liye. Prompt injection agent se aisa code likhwa sakta hai jo data bahar bheje; network band ho to yeh raasta hi nahi.

**Q4. Blast radius kya hai?**
> Kuch ghalat ho to nuqsan kitna phail sakta hai. Sandboxing, least privilege aur ephemeral environments se ise chhota rakhte hain.

## ⚠️ Aam galtiyan
- Demo mein `exec()` se chalana aur wahi production mein le jana
- Sandbox mein API keys env variables ki shakal mein de dena
- Timeout na lagana
- Ek hi sandbox kayi users mein share karna

## ✅ Yaad rakhne wali baat
> **Model ka code = ajnabi ka code.** Alag rakho, band rakho, chhota rakho, aur kaam ke baad mita do.

# 5.9 Agent Reliability, UX & Failure Recovery — Bharose ke qabil agent

🎯 **Ek line mein:** Acha agent woh nahi jo kabhi fail na ho — balki woh jo **predictable tareeqe se fail ho, gracefully sambhle, aur user ko informed aur in-control rakhe.**

## 🧠 Common Agent Failures — symptom → wajah → ilaj

| Failure | Aam wajah | Ilaj |
|---|---|---|
| **Ghalat tool chuna** | Milte julte tools, kamzor descriptions | Overlap khatam, behtar descriptions, kam tools |
| **Ghalat arguments** | Loose schema, model ne andaza lagaya | Enums/types, validation, saaf error wapas |
| **Tool exist nahi karta (hallucinated tool)** | Model ne naam khud bana liya | Strict schema, unknown tool par error observation |
| **Dohrai actions** | Retry par asli duniya mein dobara action | **Idempotency keys** |
| **Infinite loop** | Ghair-wazeh errors, koi progress check nahi | Max steps, repeat detection, behtar errors |
| **Adhoora kaam / jaldi exit** | Model ne samjha kaam ho gaya | Completion criteria, verification step |
| **Context ka phailna** | Bare tool outputs, lambi history | Truncation, summary (L6) |
| **Goal se bhatakna** | Lambe runs | Plan/goal ko state mein pin, progress notes |

## 💰 Budgets — har agent ki "credit limit"
- **Max steps** (iterations)
- **Timeout** (wall-clock waqt)
- **Retry budget** (kitni bar dobara)
- **Token / cost budget** per request ya per user
- Budget khatam → ruko, **jo ho chuka woh batao**, aur insaan ya user ko agla raasta do

## 🔁 Idempotent Operations — "do bar payment nahi"
Network fail hua, agent ne refund dobara call kar diya → customer ko **do refund**. 😱
- Har consequential action ke saath ek **idempotency key** (jaise `refund-order1043-attempt1`)
- Server same key dobara dekhe to **naya action nahi**, purana nateeja wapas
- Isi tarah: "pehle check karo ho chuka hai ya nahi, phir karo"

## 💾 Persisting State to Resume
Lamba agent 20 minute chala aur crash — shuru se nahi, **aakhri checkpoint se** (5.7). Kaam ke durable records: progress file, task list, DB status.

## 🪂 Fallbacks & Human Escalation

**Fallback ki seerhi:**
1. Tool fail → retry (temporary error)
2. Doosra tool / doosra provider / doosra model
3. Agent se **workflow / fixed jawab** par degrade
4. **Insaan ko escalate**

**Escalation triggers (support bot):**
- Customer saaf kahe "insaan se baat karni hai"
- Naraz / ghusse wala lehja
- Paisa ek had se zyada
- Agent 2–3 bar ek hi kaam mein fail
- Model ka **confidence kam** / policy mein jawab nahi
- Qanooni ya hassas mamla

**Handoff acha kaise ho:** insaan ko **poora khulasa** — customer kaun, masla kya, agent ne kya try kiya — taake customer ko sab dobara na batana pare.

## 🖥️ UX: User ko kya dikhana hai (Surfacing Progress)
- **Abhi kya ho raha hai:** "Aap ka order check kar raha hoon…" (WhatsApp par *typing* ya chhota status)
- **Sources / saboot:** jawab kis data par mabni hai
- **Approval requests saaf:** "Main Rs 2,500 ka refund order #1043 par karna chahta hoon. Haan / Nahi?" — kya, kitna, kis par
- **Control:** **Cancel, Pause, Resume, Redirect** — lambe kaam mein user beech mein raasta badal sake

## 🔭 Observability (L7/L10 ka pul)
- Har run ka **trace**: har step, tool call, arguments, nateeja, latency, tokens
- **Cost per run**, steps per run, success rate
- Fail hue traces → **eval dataset** mein shamil (L7)

## 📉 Compounding error — dobara yaad
Har qadam 95% → 10 qadam ~60%. **Reliability ka raaz:** kam steps, mazboot tools, verification, checkpoints aur human gates.

## 🎤 Interview Q&A — 5.9

**Q1. Production agent ki reliability kaise barhayenge?**
> Steps kam aur tools mazboot (typed schemas, saaf errors), budgets aur stopping conditions, idempotent consequential actions, checkpointing for resume, fallbacks aur human escalation, aur trace-based observability ke saath trajectory evals. Har production failure ko eval case bana deta hoon.

**Q2. Agent ne retry par customer ko do bar refund kar diya. Kya ghalat hua, kaise rokenge?**
> Consequential action idempotent nahi tha. Har action par idempotency key, server-side duplicate check, "pehle status check phir action" pattern, aur refund par human approval. Retries sirf temporary errors par aur code level par, model ke faislay par nahi.

**Q3. 30 minute ka agent run beech mein crash ho gaya. Design kaisa hona chahiye tha?**
> Har step ke baad state checkpoint, durable progress records (task list/progress file), aur resume logic taake aakhri mukammal step se jari ho. Side-effect wale steps idempotent taake dobara chalna mehfooz ho.

**Q4. Agent ko insaan ke paas kab escalate karna chahiye?**
> User ki request par, naraz customer, paisa had se zyada, bar bar failure, kam confidence, ya policy se bahar sawal. Aur handoff ke saath poora context summary.

**Q5. Lambe agent kaam mein user ko kya dikhana chahiye?**
> Live progress, sources, saaf approval requests (kya, kitna, kis par), aur cancel/pause/redirect ka control.

## ⚠️ Aam galtiyan
- Sirf "happy path" test karna
- Retry logic model ke haath mein dena
- Escalation mein insaan ko context na dena
- Bina trace ke production mein chalana ("pata hi nahi kya hua")

## ✅ Yaad rakhne wali baat
> **Predictable failure > hidden failure.** Budgets, idempotency, checkpoints, fallbacks, escalation aur visibility.

# 5.10 Agent Frameworks & Libraries — Ek agent, kayi frameworks

🎯 **Ek line mein:** Framework **popularity** se nahi, **architecture ki zaroorat** se chuno — aur neeche ka agent loop har haal mein khud samjho.

## 🧠 Bare frameworks — ek nazar mein

| Framework | Andaaz | Taqat | Kamzori / Dhyan | Best for |
|---|---|---|---|---|
| **LangGraph** | Low-level graph, explicit state | Control flow, persistence, loops, HITL, debugging | Seekhne mein waqt, zyada boilerplate | Production, complex, stateful agents |
| **Google ADK** (Agent Development Kit) | Code-first agent toolkit | Multi-agent, sessions/state, tools, deployment, A2A, Google Cloud integration | Google ecosystem ki taraf jhukao | Gemini/Vertex AI par bane systems |
| **CrewAI** | Role-based "crew" (role, goal, backstory) + **Flows** | Jaldi multi-agent prototype, samajhne mein aasan | Bareek control kam (Crews mein) | Role-based teams, tez prototypes |
| **Agno** (pehle *Phidata*) | Agents, Teams, Workflows + runtime | Tez, halka, agents ko service bana kar dena | Nisbatan naya ecosystem | Jaldi production-ready agent services |
| **OpenAI Agents SDK** | Halka, code-first | Tools, **handoffs**, guardrails, sessions, tracing built-in | OpenAI-centric defaults (doosre models bhi chal sakte) | Simple-to-medium agents, handoff patterns |
| **Claude Agent SDK** | Claude Code wala harness | Built-in tools, context management, subagents, hooks, MCP | Claude-centric | Lambe, tool-heavy agents (coding, research) |
| **Raw SDK + apna loop** | Khud likho | Poora control, koi abstraction nahi | Sab kuch khud (state, retries) | Seekhna, chhote agents, khaas zaroorat |

## 📐 Comparison Criteria — interview mein in alfaaz mein jawab do
1. **Abstraction level** — kitna chhupata hai? Debugging kitni aasan?
2. **Execution control** — flow main tay karta hoon ya framework?
3. **State & persistence** — checkpointing, resume, memory?
4. **Multi-agent support** — supervisor, handoffs, teams?
5. **Tool integrations** — MCP support, A2A?
6. **Human-in-the-loop** — approvals, interrupts?
7. **Observability** — tracing built-in ya alag tool?
8. **Deployment** — server, scaling, managed runtime?
9. **Vendor flexibility** — model badalna kitna aasan?
10. **Community & maturity** — docs, examples, stability

## 🏭 Real-world faisla (watch store bot)
- Zaroorat: tools, approval step (refund), conversation state resume, WhatsApp webhook se chalna
- **LangGraph** — kyunke HITL interrupt + checkpointing chahiye
- Ya simple version: **raw SDK loop** + DB mein state — agar flow chhota hai
- CrewAI yahan zyada mauzoo nahi — koi "team of roles" ki zaroorat nahi

## 🎤 Interview Q&A — 5.10

**Q1. Aap kaunsa agent framework use karenge aur kyun?**
> *Jawab ka dhancha:* "Depends on requirements" + ek concrete faisla. Misal: "Mujhe explicit control flow, human approval aur crash ke baad resume chahiye tha, is liye LangGraph. Agar simple handoff-based agent hota to OpenAI Agents SDK, aur role-based prototype ke liye CrewAI." Tool names nahi, **trade-offs** batao.

**Q2. Framework ke baghair agent kyun banana chahiye (kabhi)?**
> Seekhne ke liye taake loop ke mechanics samajh aayein, aur production mein jab agent simple ho — kam dependencies, poora control, aasan debugging.

**Q3. LangGraph aur CrewAI mein farq?**
> LangGraph low-level graph framework hai — explicit state aur control flow, production ke liye. CrewAI high-level role-based abstraction hai — jaldi multi-agent banana, magar bareek control kam.

## ⚠️ Aam galtiyan
- Framework seekhna, agent loop nahi
- Har naye framework par project dobara likhna
- Framework ka "magic" production mein debug na kar paana

## ✅ Yaad rakhne wali baat
> **Frameworks badalte rehte hain; agent loop, state, tools aur reliability ke usool nahi.**

Roadmap ka capstone: **"Build a Production-Style Agent"** — teen options. Yahan code nahi, **design soch** — jo interview mein project defend karte waqt kaam aati hai.

**Har option ki shartain:** kam az kam 3 kaam ke tools · multi-step autonomous execution · explicit state · stopping conditions · failure recovery · logs aur progress visibility · consequential action par kam az kam ek human approval · framework mein banao magar loop first principles se samjha sako.

## 🅰️ Option A — Research Agent

```
 User question
      │
      ▼
 [Planner] → sub-questions (3–6)
      │
      ▼
 [Researchers] (parallel) → web_search, fetch_page, read_pdf
      │  har ek: notes + sources
      ▼
 [Analyst] → takrao, kamzor saboot, khala dhoondo → zaroorat ho to re-plan
      │
      ▼
 [Writer] → cited report
      │
      ▼
 [HITL] → "Report publish/email karun?" approve
```
- **State:** question, plan, notes[], sources[], open_gaps, draft
- **Stop:** sab sub-questions answered ya budget khatam ya naye sources kuch naya nahi de rahe

- **HITL:** bahar email/message bhejne se pehle approval
- **Interview point:** "Server reusable hai — kal koi aur MCP-compatible app bhi isse jur sakti hai; tools least privilege ke saath."

## 🅲 Option C — Coding / Data Agent
```
 Task samjho → plan → code likho → [SANDBOX] chalao
        ▲                               │
        └──── error parh kar theek ◄────┘ (max N attempts)
                                        │
                                    nateeja / chart / file
```
- **State:** task, plan, attempts, last_error, artifacts
- **Stop:** tests/validation pass, ya max attempts
- **HITL:** kisi asli file/DB ko badalne se pehle
- **Interview point:** "Reflection yahan sab se zyada faida deti hai kyunke error ek objective feedback hai; sandbox se blast radius chhota."

# 12 · Case Study: WhatsApp Support Agent ka poora design

Yeh section poore Level 5 ko **ek asli system** mein jorta hai. Interview mein "Design a customer support agent" aaye to yahi dhancha.

## 1️⃣ Requirements
- Customers WhatsApp par sawal karte hain: order status, delivery, COD, exchange, product sawal, shikayat
- Jawab tez (chand seconds), Urdu/Roman Urdu/English
- Paise wale actions mehfooz; insaan ko handoff ka raasta
- Kam cost per conversation

## 2️⃣ Workflow vs Agent ka batwara
```
 WhatsApp message
       │
       ▼
  [Router / classifier] (sasta model)
       │
   ┌───┼─────────────┬─────────────────┐
   ▼   ▼             ▼                 ▼
 FAQ  Order status  Complex case      "Insaan chahiye" / ghussa
 (RAG (workflow)    (AGENT + tools)   (foran handoff)
 on policy)
```
**Wajah:** zyada tar traffic aasan cases ka — unhein agent par chalana mehnga aur risky.

## 3️⃣ Tools (permission tiers ke saath)
| Tool | Tier | Note |
|---|---|---|
| `get_order(order_id)` | Read | customer_id **session se**, model se nahi |
| `track_shipment(tracking_no)` | Read | |
| `check_stock(product)` | Read | |
| `get_policy(topic)` | Read | RAG on policy docs |
| `create_exchange_request(...)` | Reversible write | log + customer ko confirmation |
| `update_address(...)` | Reversible write | sirf dispatch se pehle (code check) |
| `issue_refund(...)` | **Approval** | idempotency key, amount ≤ order amount |
| `handoff_to_human(summary)` | — | poora context ke saath |

## 4️⃣ Memory & State
- **Thread state:** current order_id (pinned), exchange stage, approval status
- **Long-term:** zuban ki pasand, pichli shikayaton ka khulasa
- **Memory nahi:** order status, stock (hamesha tool se live)

## 5️⃣ Loop, budgets & stopping
- ReAct loop, **max ~6 steps**, per-conversation token budget
- Same tool+args 2 bar → ruk kar clarification ya handoff

## 6️⃣ Human-in-the-loop
- Refund/cancel → owner ko approval request (kya, kitna, kis order par)
- Approval ka intezar → state checkpoint; approval aate hi resume

## 7️⃣ Guardrails & security
- Customer ke message aur docs = **data**, hukm nahi (prompt injection)
- Doosre customer ka data kabhi nahi (authorization in code)
- PII logs mein mask

## 8️⃣ Failure handling
- Shopify/API down → "abhi system slow hai, thori dair mein update deta hoon" + retry queue
- Model provider down → fallback model ya FAQ-only mode
- 2–3 failures → insaan

## 9️⃣ Naapna (metrics)
- **Resolution rate** (baghair insaan), **escalation rate**, **ghalat tool/argument rate**
- Latency (p95), **cost per conversation**, customer rating
- Fail hue conversations → eval set (L7)

> 🎯 Interview mein yeh 9 qadam is tarteeb se bolo — aap "tool names" nahi, **engineering faisle** sunaoge.

| 27 | Checkpointing ka faida? | Resume, HITL, time travel, fault tolerance |
| 28 | Sandbox kyun? | Untrusted code se host, secrets, network ki hifazat |
| 29 | Idempotency? | Retry par action dobara na ho (idempotency key) |
| 30 | Framework kaise chunein? | Control, state, HITL, multi-agent, observability, vendor flexibility |

## 🏛️ System Design sawal: "E-commerce ke liye customer support agent design karein"

**Jawab ka tarteebwar dhancha (8–10 minute):**

1. **Sawal wazeh karo (assumptions):** traffic kitna? channels? zubanein? kaunse actions allowed? latency target? budget?
2. **Baseline:** pehle router + FAQ (RAG) + order-status workflow — "yeh 70–80% handle kar lega"
3. **Agent kahan:** sirf complex multi-step cases ke liye, limited tools
4. **Tools + permission tiers:** read / reversible write / approval
5. **State & memory:** thread state (pinned IDs), long-term preferences, live data via tools
6. **Reliability:** budgets, stopping, idempotency, fallbacks, escalation triggers
7. **Security:** prompt injection (data ≠ hukm), authorization in code, PII masking
8. **HITL:** refund/cancel approval, insaan ko handoff with summary
9. **Observability & evals:** traces, cost/conversation, resolution rate, trajectory evals
10. **Scaling & trade-offs:** sasta model router par, bara model agent par; caching FAQ; "agar traffic 10x ho to…"

> 🔑 Har qadam par **"kyun"** batao aur ek **rejected alternative**: *"Maine multi-agent nahi liya kyunke…"*

## 🐞 Debugging Scenarios (real interviews mein aate hain)

**S1. "Agent ghalat tool chun raha hai."**
> Trace dekho → kis input par? Tools overlap? Description mubham? Fix: descriptions mein kab/kab-nahi, overlapping tools merge, eval set par tool-selection accuracy naapo.

**S2. "Agent loop mein phans kar paisa kha raha hai."**
> Fori: max steps + budget cap + repeat detection. Wajah: tool errors ghair-wazeh, ya completion criteria saaf nahi. Error messages ko actionable banao.

**S3. "Test mein sahi, production mein fail."**
> Production inputs mukhtalif (Roman Urdu, typos, lambi chats), real API latency/errors, context bharna. Production traces se fail cases nikal kar eval set mein daalo; failure injection testing.

**S4. "Cost achanak double ho gayi."**
> Steps per run barhe? Tool outputs bare ho gaye? Prompt lamba hua? Retry storm? Per-run cost aur steps ka monitoring, tool output truncation, budget caps.

**S5. "Customer A ko customer B ka order dikh gaya."**
> Authorization model par chhori hui thi ya memory namespace galat. Fix: identity session se, har tool call par code-level auth, memory retrieval user_id filter ke saath, aur yeh case security regression test.

## 🗣️ Interview mein bolne ka andaaz
- Pehle **assumptions**, phir **simplest solution**, phir complexity
- **Tool names kam, trade-offs zyada**
- Pata na ho to: *"Main ise aise naapunga/test karunga…"* — bluff nahi
- Numbers ke saath: "steps 6 ki had", "p95 latency", "cost per conversation"

# 15 · 🧪 Self-Check Quiz

Pehle khud jawab socho, phir **"Jawab dekho"** par click karo.

**1. Ek system mein LLM sirf yeh tay karta hai ke sawal "billing" hai ya "delivery", baqi sab code karta hai. Yeh agent hai?**
<details><summary>Jawab dekho</summary>
Nahi — yeh <b>router</b> (workflow ka hissa) hai. Agent tab hota jab model loop mein khud tools chun kar agle qadam tay karta.
</details>

**2. Har step 90% reliable ho to 5 steps ka end-to-end success takreeban kitna?**
<details><summary>Jawab dekho</summary>
0.9 ki power 5 ≈ <b>59%</b>. Is liye steps kam aur har step mazboot.
</details>

**3. Agent "dobara socho" reflection se apni ghalti kyun nahi pakar pata?**
<details><summary>Jawab dekho</summary>
Bahar ke naye saboot ke baghair model wahi soch dohrata hai. Reflection tab kaam karti hai jab tests, validators ya tool errors jaisa <b>objective feedback</b> ho.
</details>

**4. Tool ne sirf "Error 400" wapas kiya aur agent loop mein phans gaya. Asal masla?**
<details><summary>Jawab dekho</summary>
Error message <b>actionable nahi</b> — model ko pata hi nahi kya badle. Error mein batao kya ghalat hai aur agla qadam kya ho.
</details>

**5. Refund ki had prompt mein likhi hai: "5000 se zyada refund mat dena." Kaafi hai?**
<details><summary>Jawab dekho</summary>
Nahi. Prompt security boundary nahi. Had <b>code mein validate</b> karo, aur refund par human approval + idempotency key.
</details>

**6. MCP server ke teen primitives aur unka control kis ke paas?**
<details><summary>Jawab dekho</summary>
<b>Tools</b> (model), <b>Resources</b> (application), <b>Prompts</b> (user).
</details>

**7. MCP aur A2A ek doosre ke competitor hain?**
<details><summary>Jawab dekho</summary>
Nahi. MCP agent ko tools/data se jorta hai, A2A agents ko agents se. Dono saath chalte hain.
</details>

**8. Order ka live status long-term memory mein save karna chahiye?**
<details><summary>Jawab dekho</summary>
Nahi — yeh live data hai, stale ho jayega. Har bar <b>tool</b> se taaza lao. Memory mein durable facts aur preferences.
</details>

**9. Human approval ka intezar 3 ghante chala. LangGraph mein kaunsi cheez ise mumkin banati hai?**
<details><summary>Jawab dekho</summary>
<b>Checkpointer</b> (thread_id ke saath state save) + <b>interrupt</b>. Approval aate hi graph usi checkpoint se resume.
</details>

**10. Multi-agent research system ka sab se bara nuqsan?**
<details><summary>Jawab dekho</summary>
Bohot zyada <b>token cost</b> (simple chat se kai guna) aur coordination/context-sharing ke masle. Sirf qeemti, parallelizable kaamon mein faida.
</details>

**11. Sandbox mein network kyun band?**
<details><summary>Jawab dekho</summary>
<b>Data exfiltration</b> aur malicious downloads rokne ke liye — khaas kar prompt injection se likhwaye code ke khilaf.
</details>

**12. Retry par customer ko do bar refund hua. Ek lafz ka ilaj?**
<details><summary>Jawab dekho</summary>
<b>Idempotency</b> (idempotency key + server-side duplicate check).
</details>

**13. Interview mein "kaunsa framework?" ka behtareen jawab ka dhancha?**
<details><summary>Jawab dekho</summary>
Requirements (control flow, state, HITL, multi-agent, observability, vendor flexibility) → unke hisaab se ek concrete faisla → ek rejected alternative ki wajah.
</details>

# ✅ Level 5 ka Nichor (Final Summary)

1. **Agent = model khud agla qadam tay kare.** Simple se shuru karo; agent sirf jab zaroori.
2. **ReAct** soch ko saboot se jorta hai; **planning** lambe kaam ke liye; **reflection** bahar ke feedback ke saath.
3. **Tools** agent ki taqat bhi hain aur khatra bhi — achhi description, typed params, actionable errors, **permission tiers**.
4. **MCP** tools jorne ka standard; har server ek trust decision.
5. **Memory:** History ≠ State ≠ Memory; Write-Retrieve-Update-Forget; live data tools se.
6. **Multi-agent** tabhi jab specialization ka faida complexity se zyada.
7. **LangGraph:** state + nodes + edges + checkpointer → wazeh flow, resume, HITL.
8. **Sandbox:** model ka code ajnabi ka code; blast radius chhota.
9. **Reliability:** budgets, idempotency, checkpoints, fallbacks, escalation, visibility.
10. **Frameworks** badalte hain; usool nahi.

### ➡️ Agla Level
**Level 6 · Context Engineering** — jo kuch yahan "context bharna", "tool outputs bare", "summary mein ID khona" ke tor par aaya, woh wahan gehrai se: context ko budget ki tarah manage karna aur lambe agents ke liye **harness** banana.

> *"Videos dekhna kaafi nahi — level tab mukammal jab project ban jaye aur aap yeh sab concepts kisi dost ko apne alfaaz mein samjha sakein."*
