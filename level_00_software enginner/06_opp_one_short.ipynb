# Static Variables(Vs Instance variables)

1. Instance Variable

Instance variable har object (instance) ki apni alag copy hoti hai. Agar ek object ki value change karo, to dusre object par koi effect nahi hota.

```python
class Student:

    def __init__(self, name):
        self.name = name      # Instance Variable


s1 = Student("Ali")
s2 = Student("Ahmed")

print(s1.name)
print(s2.name)
```

```python
# if
s1.name = "Usman"

print(s1.name)
print(s2.name)
```

✅ Sirf s1 ki value change hui.

2. Static Variable (Class Variable)

Static variable class ke andar define hoti hai aur sab objects ke liye common (shared) hoti hai.

```python
class Student:

    school = "ABC School"      # Static Variable

    def __init__(self, name):
        self.name = name       # Instance Variable


s1 = Student("Ali")
s2 = Student("Ahmed")

print(s1.school)
print(s2.school)
```

```python
# Static Variable Change Karna
Student.school = "XYZ School"

print(s1.school)
print(s2.school)
```

✅ Dono objects ki value change ho gayi kyun ke ye class variable hai.

# interview diffrence


| Instance Variable                    | Static Variable                                     |
| ------------------------------------ | --------------------------------------------------- |
| `self.variable` se banti hai         | Class ke andar directly banti hai                   |
| Har object ki apni copy hoti hai     | Sab objects ke liye ek hi copy hoti hai             |
| Object ke through access hoti hai    | Class ya object dono ke through access ho sakti hai |
| Object-specific data store karti hai | Common data store karti hai                         |
| Example: `name`, `age`, `salary`     | Example: `company`, `school`, `country`             |


## Points to Remember About Static Variables
- Static attributes class level par create hote hain, object ke andar nahi.
- Static attributes ko ClassName.attribute_name ke through access karna best practice hai.
- Static attributes object se independent hote hain. Inhein access karne ke liye object (instance) banana zaroori nahi hota.
- Static attribute ki value class ke tamam objects ke darmiyan shared hoti hai. Agar value change ho jaye, to us class ke sab objects par us ka asar hota hai.

```python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```

# Inheritance in Python (Roman Urdu)
Inheritance OOP ka ek concept hai jisme ek class (Child/Subclass) doosri class (Parent/Superclass) ki properties (variables) aur methods ko inherit (reuse) karti hai.

Iska main purpose:

- Code Reusability
- Code Duplication kam karna
- Easy Maintenance

## Types of Inheritance

Python me 5 types ki inheritance hoti hain.

- single Inheritance

Ek child sirf ek parent se inherit kare.
- Multiple Inheritance

Ek child multiple parents se inherit kare.l(like two parent)

- Multilevel Inheritance

Ek class doosri se inherit kare aur teesri us se inherit kare.

- Hierarchical Inheritance

Ek parent aur us se multiple children.

- Hybrid Inheritance

Hybrid inheritance multiple inheritance ka combination hoti hai.

```python
# single inheritance
class User:

  def __init__(self):
    self.name = 'nitish'
    self.gender = 'male'

  def login(self):
    print('login')

# child
class Student(User):

  def __init__(self):
    super().__init__() # Call the parent's __init__ method
    self.rollno = 100

  def enroll(self):
    print('enroll into the course')

u = User()
s = Student()

print(s.name)
s.login()
s.enroll()
```

```python
# Multiple Inheritance

# parent
class User:

  def __init__(self):
    self.name = 'nitish'
    self.gender = 'male'

  def login(self):
    print('login')

# child
class Student(User):

  def __init__(self):
    super().__init__() # Call the parent's __init__ method
    self.rollno = 100

  def enroll(self):
    print('enroll into the course')

u = User()
s = Student()

print(s.name)
s.login()
s.enroll()
```

```python
# Multilevel Inheritance

class GrandParent:

    def house(self):
        print("Grandparent House")


class Parent(GrandParent):

    def car(self):
        print("Parent Car")


class Child(Parent):

    def bike(self):
        print("Child Bike")


c = Child()

c.house()
c.car()
c.bike()
```

```python
# Hierarchical Inheritance

class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Bark")


class Cat(Animal):

    def meow(self):
        print("Meow")


d = Dog()
c = Cat()

d.eat()
d.bark()

c.eat()
c.meow()
```

```python
# Hybrid Inheritance
class A:

    def showA(self):
        print("A")


class B(A):

    def showB(self):
        print("B")


class C(A):

    def showC(self):
        print("C")


class D(B, C):

    def showD(self):
        print("D")


obj = D()

obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

#### Diamond Problem fix in hybrid inheritance

Python is problem ko kaise solve karta hai?

Python MRO (Method Resolution Order) use karta hai.

```python
class A:
    def show(self):
        print("A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

obj = D()
obj.show()
```

## Method Overriding
Jab child class parent class ke method ko same name aur same parameters ke sath dobara likhti (redefine) hai, usay Method Overriding kehte hain.

```python
class Phone:

    def __init__(self, price, brand):
        self.price = price
        self.brand = brand

    def buy(self):
        print("Buying a Phone")


class SmartPhone(Phone):

    def buy(self):          # Overriding Parent Method
        print("Buying a SmartPhone")


s = SmartPhone(50000, "Apple")

s.buy()
```

#Abstraction

Abstraction ka matlab hai sirf important details user ko dikhana aur implementation (internal logic) ko hide karna.

Real-life Example:

- `ATM:` Tum sirf PIN enter karte ho aur paisay nikalte ho. Andar transaction kaise process hoti hai, wo tumhein nahi dikhai deta.
- `Car:` Tum steering, brake aur accelerator use karte ho. Engine andar kaise kaam karta hai, ye hidden hota hai.

Python me Abstraction

Python me abstraction ke liye abc module use hota hai.

Isme do cheezen hoti hain:

- ABC → Abstract Base Class
- @abstractmethod → Abstract Method

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car Started")


c = Car()

c.start()
```
