## TOLL CALLING

```python
def get_weather(city: str) -> str:
    """Get the current weather conditions for a specific city."""
    weather = {"Tokyo": "65°F, clear skies", "London": "50°F, rainy"}
    return weather.get(city, "Weather data unavailable.")

## ------------------ chat message -------------------------------------


messages = [{
    "role": "user",
    "content": "What are current weather conditions in capital of Japan"
}]

response = ollama.chat(
    model='qwen3:8b',
    messages=messages,
    tools=[get_weather],
    think=False
)
messages.append(response.message)
```

```python
response.message.tool_calls
```

```python
# TOOL CALLING
tools = {
    "get_weather":get_weather
}

tool_call = response.message.tool_calls[0]
tool_name = tool_call.function.name
arguments = tool_call.function.arguments

# Call the corresponding Python function
result = tools[tool_name](**arguments)
print("Tool Output:", result)
print("****************************"*10)

# CALLING LLM 
messages.append({
                "role": "tool",
                "name": tool_name,
                "content": str(result)
            })


response = ollama.chat(
    model='qwen3:8b',
    messages=messages,
    tools=[get_weather],
    think=False
)

print(response.message.content)
```
