```python
import pypdf
import pymupdf4llm
```

```python
PDF_PATH = "dummy_book.pdf"
```

## LOADING PDF WITHOUT MARKDOWN FILE.

```python
# =====================================================================
# STAGE 1: Standard pypdf Loader (Raw Text)
# =====================================================================
def load_with_pypdf(path: str) -> str:
    extracted_text = ""
    with open(path, "rb") as file:
        reader = pypdf.PdfReader(file)
        for page in reader.pages:
            # extract_text often loses paragraph breaks and header formatting
            extracted_text += page.extract_text() + "\n"
    return extracted_text
```

```python
document1= load_with_pypdf(PDF_PATH)
```

```python
print(document1)
```

```python
# Save it to a file to look at it
with open("output_preview_1.md", "w", encoding="utf-8") as f:
    f.write(document1)

print("Saved Markdown to output_preview.md!")
```

## LOADING PDF AS A MARKDOWN FILE.

```python
# =====================================================================
# STAGE 2: pymupdf4llm Loader (Markdown)
# =====================================================================
def parse_pdf_to_markdown(path: str) -> str:
    return pymupdf4llm.to_markdown(path)
```

```python
parsed_document =  parse_pdf_to_markdown(PDF_PATH)
```

```python
print(parsed_document)
```

```python
# Save it to a file to look at it
with open("output_preview_1.md", "w", encoding="utf-8") as f:
    f.write(parsed_document)

print("Saved Markdown to output_preview.md!")
```

```python
from langchain_community.document_loaders import AsyncHtmlLoader
from langchain_community.document_transformers import Html2TextTransformer

# ==========================================
# FUNCTION 1: Load RAW HTML (No Conversion)
# ==========================================
def load_raw_html(url: str):
    """
    Fetches the webpage and keeps all the raw HTML tags (<head>, <body>, <div>, etc.)
    """
    print(f"Fetching RAW HTML from {url}...")
    loader = AsyncHtmlLoader([url]) #It downloads HTML code from websites asynchronously
    docs = loader.load()
    return docs


raw_results= load_raw_html("https://lilianweng.github.io/posts/2023-06-23-agent/")
```

```python
print(raw_results[0])
```

```python
print(raw_results[0])
```

```python
# ==========================================
# FUNCTION 2: Load HTML and Convert to MARKDOWN
# ==========================================
def load_and_convert_to_markdown(url: str):
    """
    Fetches the webpage and immediately converts the HTML tags into clean Markdown.
    """
    print(f"Fetching and converting to MARKDOWN from {url}...")
    loader = AsyncHtmlLoader([url])
    raw_docs = loader.load()
    
    # Initialize the transformer and convert the documents
    html2text = Html2TextTransformer()
    markdown_docs = html2text.transform_documents(raw_docs)
    
    return markdown_docs


raw_results_md= load_and_convert_to_markdown("https://lilianweng.github.io/posts/2023-06-23-agent/")
```

```python
print(raw_results_md[0])
```

```python
filename = "output_preview_html.md"

# Save it to a file to look at it
with open(filename, "w", encoding="utf-8") as f:
    # Use .page_content to extract the string from the Document
    f.write(raw_results_md[0].page_content)  

print(f"Saved Markdown to {filename}!")
```
